```python
# CELL 1: IMPORT LIBRARIES AND DEFINE PATHS

import pandas as pd
import os

# Project root = one level above the notebooks folder
BASE_PATH = ".."

RAW_PATH = os.path.join(BASE_PATH, "data", "raw")

TRANS_PATH = os.path.join(RAW_PATH, "LI-Small_Trans.csv")
ACCOUNTS_PATH = os.path.join(RAW_PATH, "LI-Small_accounts.csv")
PATTERNS_PATH = os.path.join(RAW_PATH, "LI-Small_Patterns.txt")

print("Raw data path:", RAW_PATH)
```

```python
# CELL 2: LOAD AND INSPECT LI-SMALL TRANSACTIONS

trans_df = pd.read_csv(TRANS_PATH)

print("TRANSACTION DATASET")
print("=" * 60)

print("Shape:", trans_df.shape)

print("\nColumns:")
for i, col in enumerate(trans_df.columns, 1):
    print(f"{i}. {col}")

print("\nData Types:")
print(trans_df.dtypes)

print("\nFirst 5 rows:")
display(trans_df.head())

print("\nMissing Values:")
print(trans_df.isnull().sum())

print("\nBasic Statistics:")
display(trans_df.describe(include="all").T)
```

```python
# CELL 3: LOAD AND INSPECT LI-SMALL ACCOUNTS

accounts_df = pd.read_csv(ACCOUNTS_PATH)

print("ACCOUNTS DATASET")
print("=" * 60)

print("Shape:", accounts_df.shape)

print("\nColumns:")
for i, col in enumerate(accounts_df.columns, 1):
    print(f"{i}. {col}")

print("\nData Types:")
print(accounts_df.dtypes)

print("\nFirst 5 rows:")
display(accounts_df.head())

print("\nMissing Values:")
print(accounts_df.isnull().sum())
# CELL 4: INSPECT LI-SMALL PATTERNS

with open(PATTERNS_PATH, "r", encoding="utf-8") as f:
    pattern_lines = f.readlines()

print("PATTERNS DATASET")
print("=" * 60)

print("Total lines:", len(pattern_lines))

print("\nFirst 40 lines:")
for i, line in enumerate(pattern_lines[:40], 1):
    print(f"{i}: {line.rstrip()}")

# Find pattern types
pattern_headers = [
    line.strip()
    for line in pattern_lines
    if line.startswith("BEGIN LAUNDERING ATTEMPT")
]

print("\nNumber of laundering pattern blocks:", len(pattern_headers))

print("\nPattern headers found:")
for pattern in pattern_headers[:20]:
    print(pattern)
print("\nUnique Values:")
for col in accounts_df.columns:
    print(f"{col}: {accounts_df[col].nunique():,}")
```

```python
# CELL 4: INSPECT LI-SMALL PATTERNS

with open(PATTERNS_PATH, "r", encoding="utf-8") as f:
    pattern_lines = f.readlines()

print("PATTERNS DATASET")
print("=" * 60)

print("Total lines:", len(pattern_lines))

print("\nFirst 40 lines:")
for i, line in enumerate(pattern_lines[:40], 1):
    print(f"{i}: {line.rstrip()}")

# Find pattern types
pattern_headers = [
    line.strip()
    for line in pattern_lines
    if line.startswith("BEGIN LAUNDERING ATTEMPT")
]

print("\nNumber of laundering pattern blocks:", len(pattern_headers))

print("\nPattern headers found:")
for pattern in pattern_headers[:20]:
    print(pattern)
```

```python
# CELL 5: ACCOUNT AND BANK IDENTIFIER INSPECTION

print("ACCOUNT IDENTIFIER INSPECTION")
print("=" * 60)

# ---------------------------------------------------------
# 1. Account Number uniqueness
# ---------------------------------------------------------

account_unique = accounts_df["Account Number"].nunique()
account_rows = len(accounts_df)

print(f"Total account records: {account_rows:,}")
print(f"Unique Account Numbers: {account_unique:,}")
print(f"Duplicate Account Number records: {account_rows - account_unique:,}")

# ---------------------------------------------------------
# 2. (Bank ID, Account Number) uniqueness
# ---------------------------------------------------------

bank_account_unique = accounts_df[
    ["Bank ID", "Account Number"]
].drop_duplicates().shape[0]

print(f"\nUnique (Bank ID, Account Number): {bank_account_unique:,}")

if bank_account_unique == account_rows:
    print("RESULT: (Bank ID, Account Number) uniquely identifies an account.")
else:
    print("RESULT: (Bank ID, Account Number) still contains duplicates.")

# ---------------------------------------------------------
# 3. Transaction → Accounts bank matching
# ---------------------------------------------------------

transaction_from_banks = set(trans_df["From Bank"].astype(str))
transaction_to_banks = set(trans_df["To Bank"].astype(str))

account_bank_ids = set(accounts_df["Bank ID"].astype(str))

from_matches = transaction_from_banks.intersection(account_bank_ids)
to_matches = transaction_to_banks.intersection(account_bank_ids)

print("\nTRANSACTION → ACCOUNT BANK MATCHING")
print("-" * 60)

print(
    f"Unique From Bank IDs in transactions: "
    f"{len(transaction_from_banks):,}"
)

print(
    f"Matched From Bank IDs: "
    f"{len(from_matches):,}"
)

print(
    f"Unique To Bank IDs in transactions: "
    f"{len(transaction_to_banks):,}"
)

print(
    f"Matched To Bank IDs: "
    f"{len(to_matches):,}"
)

# ---------------------------------------------------------
# 4. Bank identifier inspection
# ---------------------------------------------------------

unique_bank_names = accounts_df["Bank Name"].nunique()
unique_bank_ids = accounts_df["Bank ID"].nunique()
unique_bank_pairs = accounts_df[
    ["Bank Name", "Bank ID"]
].drop_duplicates().shape[0]

print("\nBANK IDENTIFIER INSPECTION")
print("-" * 60)

print(f"Unique Bank Names: {unique_bank_names:,}")
print(f"Unique Bank IDs: {unique_bank_ids:,}")
print(f"Unique Bank Name + Bank ID pairs: {unique_bank_pairs:,}")

# ---------------------------------------------------------
# 5. Check whether one Bank ID has multiple Bank Names
# ---------------------------------------------------------

bank_id_name_counts = (
    accounts_df.groupby("Bank ID")["Bank Name"]
    .nunique()
)

multiple_names_per_id = bank_id_name_counts[
    bank_id_name_counts > 1
]

print(
    f"\nBank IDs with multiple Bank Names: "
    f"{len(multiple_names_per_id):,}"
)

# ---------------------------------------------------------
# 6. Final conclusion
# ---------------------------------------------------------

print("\nFINAL IDENTIFIER CONCLUSION")
print("=" * 60)

print("Account identity:")
print("(Bank ID, Account Number)")

print("\nBank identity:")
print("Bank ID")

print("\nBank Name:")
print("Descriptive attribute, NOT the primary bank identifier.")
```
