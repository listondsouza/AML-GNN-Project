```python
import pandas as pd
import os

BASE_PATH = ".."
RAW_PATH = os.path.join(BASE_PATH, "data", "raw")
PROCESSED_PATH = os.path.join(BASE_PATH, "data", "processed")

ACCOUNTS_PATH = os.path.join(RAW_PATH, "LI-Small_accounts.csv")

accounts_df = pd.read_csv(ACCOUNTS_PATH)

print("Accounts:", accounts_df.shape)
print(accounts_df.head())
```

```python
# Check required columns
required_columns = [
    "Bank Name",
    "Bank ID",
    "Account Number",
    "Entity ID",
    "Entity Name"
]

missing = [col for col in required_columns if col not in accounts_df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

# Create collision-safe account ID
accounts_df["Account ID"] = (
    accounts_df["Bank ID"].astype(str)
    + "_"
    + accounts_df["Account Number"].astype(str)
)

# Reorder columns
accounts_df = accounts_df[
    [
        "Account ID",
        "Bank ID",
        "Account Number",
        "Entity ID",
        "Entity Name",
        "Bank Name"
    ]
]

print(accounts_df.head(10))
```

```python
print("Shape:", accounts_df.shape)

print("\nColumns:")
print(accounts_df.columns.tolist())

print("\nMissing values:")
print(accounts_df.isnull().sum())

print("\nUnique Account IDs:", accounts_df["Account ID"].nunique())
print("Duplicate Account IDs:", accounts_df["Account ID"].duplicated().sum())

print("\nUnique Bank IDs:", accounts_df["Bank ID"].nunique())
print("Unique Account Numbers:", accounts_df["Account Number"].nunique())

# Validation
assert accounts_df["Account ID"].duplicated().sum() == 0

print("\nAccount ID validation passed.")
```

```python
OUTPUT_PATH = os.path.join(
    PROCESSED_PATH,
    "accounts.csv"
)

accounts_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Saved:", OUTPUT_PATH)
```
