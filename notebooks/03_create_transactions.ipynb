```python
# CELL 1: LOAD RAW TRANSACTIONS

import pandas as pd
import os

BASE_PATH = ".."

RAW_PATH = os.path.join(
    BASE_PATH, "data", "raw"
)

PROCESSED_PATH = os.path.join(
    BASE_PATH, "data", "processed"
)

TRANS_PATH = os.path.join(
    RAW_PATH, "LI-Small_Trans.csv"
)

os.makedirs(PROCESSED_PATH, exist_ok=True)

trans_df = pd.read_csv(TRANS_PATH)

print("Raw transaction dataset loaded.")
print("Shape:", trans_df.shape)

display(trans_df.head())
```

```python
# CELL 2: VALIDATE TRANSACTION DATA AND CREATE TRANSACTION ID

expected_columns = [
    "Timestamp",
    "From Bank",
    "Account",
    "To Bank",
    "Account.1",
    "Amount Received",
    "Receiving Currency",
    "Amount Paid",
    "Payment Currency",
    "Payment Format",
    "Is Laundering"
]

# Check columns
missing_columns = [
    col for col in expected_columns
    if col not in trans_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

# Reorder columns exactly as required
trans_df = trans_df[expected_columns].copy()

# Create transaction ID
trans_df.insert(
    0,
    "Transaction ID",
    range(1, len(trans_df) + 1)
)

print("Columns validated.")
print("Transaction IDs created.")

print("\nShape:", trans_df.shape)

print("\nDuplicate Transaction IDs:",
      trans_df["Transaction ID"].duplicated().sum())

display(trans_df.head())
```

```python
# CELL 3: VALIDATE FINAL TRANSACTIONS DATASET

print("TRANSACTIONS DATASET VALIDATION")
print("=" * 60)

print("Rows:", f"{len(trans_df):,}")
print("Columns:", len(trans_df.columns))

print("\nMissing values:")
print(trans_df.isnull().sum())

print("\nClass distribution:")
print(
    trans_df["Is Laundering"]
    .value_counts()
    .rename(index={
        0: "Normal",
        1: "Laundering"
    })
)

print("\nUnique Transaction IDs:",
      trans_df["Transaction ID"].nunique())

print("\nExpected rows:",
      f"{len(trans_df):,}")

print("\nExpected columns:")
for i, col in enumerate(trans_df.columns, 1):
    print(f"{i}. {col}")
```

```python
# CELL 4: SAVE PROCESSED TRANSACTIONS DATASET

OUTPUT_PATH = os.path.join(
    PROCESSED_PATH,
    "transactions.csv"
)

trans_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Transactions dataset saved successfully.")
print("\nLocation:")
print(OUTPUT_PATH)

print("\nFinal shape:")
print(trans_df.shape)
```
